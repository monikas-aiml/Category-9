```python
# MEDICAL DIAGNOSIS - BAYESIAN REASONING

def bayes_diagnosis(fever, cough):

    prior_disease = 0.10
    prior_no_disease = 0.90

    # Likelihood probabilities
    p_fever_disease = 0.80
    p_cough_disease = 0.70

    p_fever_no_disease = 0.20
    p_cough_no_disease = 0.30

    # Naive Bayes: symptoms are assumed conditionally independent
    p_evidence_disease = (
        p_fever_disease * p_cough_disease
    )

    p_evidence_no_disease = (
        p_fever_no_disease * p_cough_no_disease
    )

    numerator = prior_disease * p_evidence_disease

    denominator = (
        numerator +
        prior_no_disease * p_evidence_no_disease
    )

    posterior = numerator / denominator

    print("Medical Diagnosis")
    print("Disease Probability:", round(posterior * 100, 2), "%")

    if posterior >= 0.50:
        print("Result: Disease suspected")
    else:
        print("Result: Disease less likely")


# Sample Input
bayes_diagnosis(True, True)
```

```python
# AUTONOMOUS VEHICLE - BAYESIAN REASONING

def accident_prediction(obstacle, poor_visibility):

    prior_accident = 0.05
    prior_safe = 0.95

    p_obstacle_accident = 0.90
    p_visibility_accident = 0.80

    p_obstacle_safe = 0.20
    p_visibility_safe = 0.30

    likelihood_accident = (
        p_obstacle_accident * p_visibility_accident
    )

    likelihood_safe = (
        p_obstacle_safe * p_visibility_safe
    )

    numerator = prior_accident * likelihood_accident

    denominator = (
        numerator +
        prior_safe * likelihood_safe
    )

    probability = numerator / denominator

    print("Autonomous Vehicle Analysis")
    print("Accident Risk:",
          round(probability * 100, 2), "%")

    if probability >= 0.50:
        print("Action: Emergency braking")
    else:
        print("Action: Continue with caution")


# Sample Input
accident_prediction(True, True)
```

```python
def triangular(x, a, b, c):
    if x <= a or x >= c:
        return 0

    if x == b:
        return 1

    if x < b:
        return (x - a) / (b - a)

    return (c - x) / (c - b)


def fuzzy_output(rules):
    numerator = 0
    denominator = 0

    for strength, output in rules:
        numerator += strength * output
        denominator += strength

    return numerator / denominator if denominator else 0


def ev_fuzzy(temperature, range_drop):

    temp_low = triangular(temperature, 0, 20, 35)
    temp_normal = triangular(temperature, 25, 35, 45)
    temp_high = triangular(temperature, 40, 55, 70)

    drop_low = triangular(range_drop, 0, 5, 15)
    drop_medium = triangular(range_drop, 10, 25, 40)
    drop_high = triangular(range_drop, 30, 50, 60)

    rules = [
        (min(temp_high, drop_high), 90),
        (min(temp_normal, drop_medium), 50),
        (min(temp_low, drop_low), 10)
    ]

    risk = fuzzy_output(rules)

    print("EV Battery Risk:", round(risk, 2))

    if risk >= 70:
        print("Battery Condition: Critical")
    elif risk >= 35:
        print("Battery Condition: Moderate")
    else:
        print("Battery Condition: Good")


# Input
ev_fuzzy(52, 42)
```

```python
def triangular(x, a, b, c):
    if x <= a or x >= c:
        return 0

    if x == b:
        return 1

    if x < b:
        return (x - a) / (b - a)

    return (c - x) / (c - b)


def fuzzy_output(rules):
    numerator = 0
    denominator = 0

    for strength, output in rules:
        numerator += strength * output
        denominator += strength

    return numerator / denominator if denominator else 0


def machine_fuzzy(vibration, temperature):

    vibration_low = triangular(vibration, 0, 1, 4)
    vibration_medium = triangular(vibration, 2, 5, 8)
    vibration_high = triangular(vibration, 6, 9, 12)

    temp_low = triangular(temperature, 20, 35, 55)
    temp_normal = triangular(temperature, 45, 65, 80)
    temp_high = triangular(temperature, 70, 90, 110)

    rules = [
        (min(vibration_high, temp_high), 95),
        (min(vibration_medium, temp_normal), 50),
        (min(vibration_low, temp_low), 10)
    ]

    risk = fuzzy_output(rules)

    print("Machine Failure Risk:", round(risk, 2))

    if risk >= 70:
        print("Action: Immediate Maintenance")
    elif risk >= 35:
        print("Action: Schedule Maintenance")
    else:
        print("Action: Machine Normal")


# Input
machine_fuzzy(9, 90)
```
